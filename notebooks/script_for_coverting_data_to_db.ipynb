```python
import pandas as pd
import numpy as np

np.random.seed(42)

# Number of rows
n = 5000

# Fake customer names (simple generator)
first_names = ["John", "Mary", "Alex", "Sarah", "David", "Linda", "James", "Sophia", "Daniel", "Emma"]
last_names = ["Smith", "Johnson", "Brown", "Williams", "Jones", "Miller", "Davis", "Garcia", "Wilson", "Taylor"]

def random_name():
    return np.random.choice(first_names) + " " + np.random.choice(last_names)

# Product catalog
products = [
    ("Laptop", 850),
    ("Smartphone", 600),
    ("Headphones", 120),
    ("Smartwatch", 200),
    ("Bluetooth Speaker", 150),
    ("Monitor", 300),
    ("Keyboard", 45),
    ("Mouse", 30),
    ("Printer", 180),
    ("External Hard Drive", 95)
]

product_names = [p[0] for p in products]
product_prices = {p[0]: p[1] for p in products}

# Store locations
stores = ["Toronto", "Calgary", "Vancouver", "Montreal", "Ottawa"]

# Generate dataset
data = {
    "order_id": np.arange(10001, 10001 + n),
    "customer_id": np.random.randint(1, 2000, n),
    "customer_name": [random_name() for _ in range(n)],
    "product": np.random.choice(product_names, n),
    "quantity": np.random.randint(1, 5, n),
    "order_date": pd.to_datetime(
        np.random.choice(pd.date_range("2024-01-01", "2026-01-01"), n)
    ),
    "store_location": np.random.choice(stores, n)
}

df = pd.DataFrame(data)

# Add price and total revenue
df["unit_price"] = df["product"].map(product_prices)
df["total_price"] = df["unit_price"] * df["quantity"]

# Preview
print(df.head())

# Save to CSV
df.to_csv("retail_dataset_5000.csv", index=False)
print("Dataset created: retail_dataset_5000.csv")
```
